```python
from Util.Problems import Problem, solution

class P028(Problem):
    number = 28
    title = "Number Spiral Diagonals"
    description = """<p>Starting with the number $1$ and moving to the right in a clockwise direction a $5$ by $5$ spiral is formed as follows:</p><p class="monospace center"><span class="red"><b>21</b></span> 22 23 24 <span class="red"><b>25</b></span><br/>
20  <span class="red"><b>7</b></span>  8  <span class="red"><b>9</b></span> 10<br/>
19  6  <span class="red"><b>1</b></span>  2 11<br/>
18  <span class="red"><b>5</b></span>  4  <span class="red"><b>3</b></span> 12<br/><span class="red"><b>17</b></span> 16 15 14 <span class="red"><b>13</b></span></p><p>It can be verified that the sum of the numbers on the diagonals is $101$.</p><p>What is the sum of the numbers on the diagonals in a $1001$ by $1001$ spiral formed in the same way?</p>"""
    spiral_size = 1001
```

### Solution notes
The top-right diagonal contains all odd squares, as each of these numbers falls on the point where exactly one square of the numbers until then has been created. Therefore, the sum of the top-right diagonal (not counting the center $1$) for a spiral of size n is equal to

$\sum_{i=2}^{\lfloor n / 2 \rfloor}(2i-1)^2$

If we multiply this result by 4, we will get an approximation of the sum of all diagonals, but it will be larger than the actual answer. We can, however, calculate how much larger. This will be dependent on the size of the current square. E.G. if we are looking at a square of size $3$, for the top-left diagonal, we counted $2$ too many. In general, for the top left diagonal, we will count the square size -1 too much. In the earlier formula, the square size was $2i-1$, therefore, we count $2i-2$ too much for the top left diagonal. For the bottom left, we count that amount twice too many, and for the bottom right three times. This results in an overestimation of $(1+2+3) \times (2i-2) = 12(i - 1)$. This means the sum for all four diagonals is equal to:

$1 + \sum_{i=2}^{\lfloor n / 2 \rfloor}4(2i-1)^2 - 12(i-1)\\
1 + \sum_{i=2}^{\lfloor n / 2 \rfloor}4(4i^2-4i+1) - 12i+12\\
1 + \sum_{i=2}^{\lfloor n / 2 \rfloor}16i^2-16i+4 - 12i+12\\
1 + \sum_{i=2}^{\lfloor n / 2 \rfloor}16i^2-28i+16$


```python
@solution(P028, first=True, make_fast=True, warmup_args=(P028.spiral_size,))
def squares_calculation(spiral_size):
    diagonal_sum = 1
    for i in range(2, (spiral_size//2) + 2):
        diagonal_sum += 16 * i ** 2 - 28 * i + 16
    return diagonal_sum
```
