```python
from Util.Problems import Problem, solution

import Util.math_functions as mathf


class P027(Problem):
    number = 27
    title = "Quadratic Primes"
    description = """<p>Euler discovered the remarkable quadratic formula:</p><p class="center">$n^2 + n + 41$</p><p>It turns out that the formula will produce $40$ primes for the consecutive integer values $0 \\le n \\le 39$. However, when $n = 40, 40^2 + 40 + 41 = 40(40 + 1) + 41$ is divisible by $41$, and certainly when $n = 41, 41^2 + 41 + 41$ is clearly divisible by $41$.</p><p>The incredible formula $n^2 - 79n + 1601$ was discovered, which produces $80$ primes for the consecutive values $0 \\le n \\le 79$. The product of the coefficients, $-79$ and $1601$, is $-126479$.</p><p>Considering quadratics of the form:</p><blockquote>
$n^2 + an + b$, where $|a| < 1000$ and $|b| \\le 1000$<br/><br/><div>where $|n|$ is the modulus/absolute value of $n$<br/>e.g. $|11| = 11$ and $|-4| = 4$</div>
</blockquote><p>Find the product of the coefficients, $a$ and $b$, for the quadratic expression that produces the maximum number of primes for consecutive values of $n$, starting with $n = 0$.</p>"""
    a_limit = 999
    b_limit = 1000
```

```python
p = P027()
p.describe()
```

### Solution notes
First the most simple brute-force solution: looping through all a and b, keeping track of the number of n for which they are prime, returning the result which stayed prime the longest.

```python
@solution(P027, first=True, max_tests= 100, make_fast=True, warmup_args=(P027.a_limit, P027.b_limit))
def brute_force(a_limit, b_limit):
    biggest_n = 0
    result = 0
    for a in range(-a_limit, a_limit + 1):
        for b in range(-b_limit, b_limit + 1):
            n = 0
            while mathf.is_prime(n ** 2 + a * n + b):
                n += 1
            if n > biggest_n:
                biggest_n = n
                result = a * b
    return result
```

```python
p.test_once("brute_force")
```

Because we start at $n=0$, $b$ always has to be prime. Otherwise, $0^2 + a \times 0 + b = b$ is already not prime.

Looking at the next case, we can narrow down what $a$ should be (where $p_1$ is a prime):

$n=1: 1^2 + a + b = p_1 \\
a = p_1 - b - 1 \\
$

As we have determined before that $b$ is prime, and all primes are odd, $odd - odd - odd = odd$ so b has to be prime and a has to be odd. Using this, we can speed up our loops.

```python
@solution(P027, best=True, make_fast=True, warmup_args=(P027.a_limit, P027.b_limit))
def reduced_scope(a_limit, b_limit):
    biggest_n = 0
    result = 0
    b_primes = mathf.prime_sieve(b_limit)
    for a in range(-a_limit, a_limit + 1, 2):
        for absolute_b in b_primes:
            for negative in range(2):
                b = -absolute_b if negative else absolute_b
                n = 0
                while mathf.is_prime(n ** 2 + a * n + b):
                    n += 1
                if n > biggest_n:
                    biggest_n = n
                    result = a * b
    return result
```

```python
p.test_all()
```
