### Solution notes
A very simple brute-force method is used for this problem, but I do not yet see any optimisations. We split the string into a list of the names and we use the "sorted" function to sort the list alphabetically. Next, we loop through each name, for each name we loop through its letters and use them to calculate the name score. Multiply this by the index of the name, and keep track of the total. Very simple. Not extremely fast, so could maybe be optimised in the future, but I'm not sure how.

```python
@solution(P022, first=True, make_fast=False, warmup_args=(P022.file_text, ))
def brute_force(text):
    names = text.replace('"','').split(",")
    alphabetical_names = sorted(names)
    total_score = 0
    for i, name in enumerate(alphabetical_names):
        name_score = 0
        for letter in name:
            name_score += ord(letter) - 64
        total_score += name_score * (i + 1)
    return total_score
```
