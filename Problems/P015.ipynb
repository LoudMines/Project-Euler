### Solution notes
In a $2 \times 2$ grid, to get from the top left to the bottom right, we need 2 moves to the **r**ight and 2 moves **d**own in any order. To calculate the total number of ways to do this, we use combinatorics: The number of ways to arrange RRDD is $\frac{4\times 3}{2} = 6$ as there are 4 places for the first R to go in the sequence of 4 moves, then 3 for the second R and the remainder is D. Generalising this leads to the following formula:

$\frac{n!-(n/2)!}{(n/2)!}$

Where $n = 2 \times$grid_size

As this function uses numbers which are too big for python, the denominator is calculated separately. Even then, the numbers are too big for numba, but as this is just a mathematical calculation, and already very fast, numba is not necessary.

```python
@solution(P015, first=True, make_fast=False, warmup_args=(P015.grid_size, ))
def mathematical_calculation(grid_size):
    n = 2 * grid_size
    denominator = 1
    while n > grid_size:
        denominator *= n
        n -= 1
    return denominator // math.factorial(grid_size)
```
