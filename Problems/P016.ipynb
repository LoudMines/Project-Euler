```python
import numpy as np
import math

from Util.Problems import Problem, solution


class P016(Problem):
    number = 16
    title = "Power Digit Sum"
    description = """<p>$2^{15} = 32768$ and the sum of its digits is $3 + 2 + 7 + 6 + 8 = 26$.</p><p>What is the sum of the digits of the number $2^{1000}$?</p>"""
    power_of_two = 1000
```

```python
p = P016()
p.describe()
```

### Solution notes
$\exp{2}{1000}$ is much too large to use as a number, so we save the number in an array and we multiply it by 2 exponent times, going digit by digit. This can probably be optimised in multiple ways, but 2 obvious optimisations are trimming the array size to the size required for storing a decimal number with the value of a binary number with exponent digits. Another optimisation is not looping through the lead zeros each time.

```python
@solution(P016, first=True, make_fast=True, warmup_args=(P016.power_of_two,))
def number_array(e):
    number_size = e // 2
    big_number = np.zeros(number_size, dtype=np.int_)
    big_number[0] = 1
    for i in range(e): # Multiply the number by 2 for each exponent
        carry = 0
        for j in range(number_size): # Loop through all digits in the final big number and multiply them by 2 (with carry)
            if big_number[j] == 0:
                big_number[j] = carry
                carry = 0
                continue
            digit_times_two = big_number[j] * 2 + carry
            if digit_times_two > 9:
                carry = (digit_times_two - (digit_times_two % 10)) // 10
            else:
                carry = 0
            big_number[j] = (digit_times_two % 10)
    return np.sum(big_number)
```

```python
p.test_all()
```

The amount of digits needed to represent a binary number of n bits in decimal is:

$floor(n \times \log10(2)) + 1$

We use this to reduce the array size. Furthermore, the carry was overcomplicated in the first algorithm, which has been improved. Finally, we no longer loop over the lead zeros and check if we need to double them, only digits which have been calculated.

```python
@solution(P016, first=True, make_fast=True, warmup_args=(P016.power_of_two,))
def optimised_number_array(e):
    number_size = math.floor(e * math.log10(2)) + 1
    big_number = np.zeros(number_size, dtype=np.int_)
    big_number[0] = 1
    calculated_digits = 1
    for i in range(e): # Multiply the number by 2 for each exponent
        carry = 0
        for j in range(number_size): # Loop through all digits in the final big number and multiply them by 2 (with carry)
            if big_number[j] == 0:
                big_number[j] = carry
                carry = 0
                continue
            digit_times_two = big_number[j] * 2 + carry
            if digit_times_two > 9:
                carry = 1
                if j + 1 > calculated_digits:
                    calculated_digits += 1
            else:
                carry = 0
            big_number[j] = (digit_times_two % 10)
    return np.sum(big_number)
```
