### Solution notes
Because Numba always makes life hard, we start with a bunch of code to turn the string into an int[][]. This would be trivial without numba, with numba this means calculating the ordinal value of each character per string and multiplying it by 10 if there is another character left, then adding that to the running total.

We then loop through the grid in a super unoptimised way, checking each possible line for each coordinate to see if there is a line of 4 numbers and if so what it's product is. This looping can be optimised a lot.

```python
@solution(P011, first=True, make_fast=True, warmup_args=(P011.grid_text,))
def brute_force(grid_text):
    grid_size = 20
    grid = np.empty((grid_size,grid_size), dtype=np.int_)
    for i, line in enumerate(grid_text.splitlines()):
        for j, number_text in enumerate(line.split()):
            running_int = 0
            for c in number_text:
                running_int = running_int * 10 + ord(c) - 48 # 48 = ord('0')
            grid[i][j] = running_int
    lines = [[1, 0], [0, 1], [1, 1], [1,-1]]
    max_product = 0
    for i in range(grid_size):
        for j in range(grid_size):
            for line in lines:
                delta_x, delta_y = line[0] * 3, line[1] * 3
                if i + delta_x >= grid_size or i + delta_x < 0 or j + delta_y >= grid_size or j + delta_y < 0:
                    continue
                numbers = [grid[i + k * line[0]][j + k * line[1]] for k in range(4)]
                product = 1
                for number in numbers:
                    product *= number
                max_product = max(max_product, product)
    return max_product
```

Re ordered some loops and added a check for 0 to optimise the brute force. This could undoubtedly still be further optimised.

```python
@solution(P011, make_fast=True, warmup_args=(P011.grid_text,))
def optimised_brute_force(grid_text):
    grid_size = 20
    grid = np.empty((grid_size,grid_size), dtype=np.int_)
    for i, line in enumerate(grid_text.splitlines()):
        for j, number_text in enumerate(line.split()):
            running_int = 0
            for c in number_text:
                running_int = running_int * 10 + ord(c) - 48 # 48 = ord('0')
            grid[i][j] = running_int
    lines = [[1, 0], [0, 1], [1, 1], [1,-1]]
    max_product = 0
    for line in lines:
        for i in range(grid_size):
            if not 0 < (i + 4 * line[0]) <= grid_size:
                continue
            for j in range(grid_size):
                if not 0 < (j + 4 * line[1]) <= grid_size:
                    continue
                numbers = [grid[i + k * line[0]][j + k * line[1]] for k in range(4)]
                if 0 in numbers:
                    continue
                product = 1
                for number in numbers:
                    product *= number
                max_product = max(max_product, product)
    return max_product
```

This uses a more optimal way to run through rows and columns by using a running totals and resetting at 0. This can be further optimised by splitting the diagonals and looping them separately so the loops can be terminated earlier when fail conditions are reached.

```python
@solution(P011, make_fast=True, warmup_args=(P011.grid_text,))
def rows_columns_first(grid_text):
    grid_size = 20
    grid = np.empty((grid_size,grid_size), dtype=np.int_)
    for i, line in enumerate(grid_text.splitlines()):
        for j, number_text in enumerate(line.split()):
            running_int = 0
            for c in number_text:
                running_int = running_int * 10 + ord(c) - 48 # 48 = ord('0')
            grid[i][j] = running_int
    max_product = 0
    for i in range(grid_size):
        row_product = 1
        col_product = 1
        row_count = 1
        col_count = 1
        for j in range(grid_size):
            if grid[i][j] == 0:
                row_count = row_product = 1
            else:
                if row_count < 4:
                    row_product = row_product * grid[i][j]
                else:
                    if grid[i][j - 4] != 0:
                        row_product = row_product // grid[i][j - 4]
                    row_product = row_product * grid[i][j]
                    max_product = max(max_product, row_product)
                row_count += 1
            if grid[j][i] == 0:
                col_count = col_product = 1
            else:
                if col_count < 4:
                    col_product = col_product * grid[j][i]
                else:
                    if grid[j - 4][i] != 0:
                        col_product = col_product // grid[j - 4][i]
                    col_product = col_product * grid[j][i]
                    max_product = max(max_product, col_product)
                col_count += 1

    diagonals = [[1, 1], [1,-1]]
    for line in diagonals:
        for i in range(grid_size):
            if not 0 < (i + 4 * line[0]) <= grid_size:
                continue
            for j in range(grid_size):
                if not 0 < (j + 4 * line[1]) <= grid_size:
                    continue
                numbers = [grid[i + k * line[0]][j + k * line[1]] for k in range(4)]
                if 0 in numbers:
                    continue
                product = 1
                for number in numbers:
                    product *= number
                max_product = max(max_product, product)
    return max_product
```

Now The diagonals have also been split into their own (many) for-loops. This greatly reduces the number of options we need to check, though it makes the code much more cumbersome and unmanagable. Though this is very large and ugly, it is the fastest method I can think of at the moment, as numba is great with for-loops. This code can probably be made a lot prettier and slightly more efficient by combining all for-loops, as They are all basically doing the same thing, but this is a lot of effort I don't want to look into right now. This might be a potential further improvement later.

```python
@solution(P011, best=True, make_fast=True, warmup_args=(P011.grid_text,))
def per_direction(grid_text):
    grid_size = 20
    grid = np.empty((grid_size,grid_size), dtype=np.int_)
    for i, line in enumerate(grid_text.splitlines()):
        for j, number_text in enumerate(line.split()):
            running_int = 0
            for c in number_text:
                running_int = running_int * 10 + ord(c) - 48 # 48 = ord('0')
            grid[i][j] = running_int
    max_product = 0
    for i in range(grid_size):
        row_product = 1
        col_product = 1
        row_count = 1
        col_count = 1
        for j in range(grid_size):
            if grid[i][j] == 0:
                row_count = row_product = 1
            else:
                if row_count < 4:
                    row_product = row_product * grid[i][j]
                else:
                    if grid[i][j - 4] != 0:
                        row_product = row_product // grid[i][j - 4]
                    row_product = row_product * grid[i][j]
                    max_product = max(max_product, row_product)
                row_count += 1
            if grid[j][i] == 0:
                col_count = col_product = 1
            else:
                if col_count < 4:
                    col_product = col_product * grid[j][i]
                else:
                    if grid[j - 4][i] != 0:
                        col_product = col_product // grid[j - 4][i]
                    col_product = col_product * grid[j][i]
                    max_product = max(max_product, col_product)
                col_count += 1

    # Diagonals to the right, if they start at a column > grid_size - 3, 4 consecutive numbers of the grid won't fit anymore.

    for start_index in range(grid_size - 3):
        row_diagonal_product = 1
        row_diagonal_count = 1
        col_diagonal_product = 1
        col_diagonal_count = 1
        # Loop through all items in the diagonal starting at this point in the row
        for other_index in range(grid_size - start_index):
            row_current_number = grid[start_index + other_index][other_index]
            if row_current_number == 0:
                row_diagonal_product = row_diagonal_count = 1
            else:
                if row_diagonal_count < 4:
                    row_diagonal_product = row_diagonal_product * row_current_number
                else:
                    if ( (0 <= start_index + other_index - 4 < grid_size) and
                            (0 <= other_index < grid_size) and
                            (grid[start_index + other_index - 4][other_index - 4] != 0)):
                        row_diagonal_product = row_diagonal_product // grid[start_index + other_index - 4][other_index - 4]
                    row_diagonal_product = row_diagonal_product * row_current_number
                    max_product = max(max_product, row_diagonal_product)
                row_diagonal_count += 1
            if start_index == 0:
                continue
            col_current_number = grid[other_index][start_index + other_index]
            if col_current_number == 0:
                col_diagonal_product = col_diagonal_count = 1
            else:
                if col_diagonal_count < 4:
                    col_diagonal_product = col_diagonal_product * col_current_number
                else:
                    if ( (0 <= start_index + other_index - 4 < grid_size) and
                            (0 <= other_index - 4 < grid_size) and
                            (grid[other_index - 4][start_index + other_index - 4] != 0)):
                        col_diagonal_product = col_diagonal_product // grid[other_index - 4][start_index + other_index - 4]
                    col_diagonal_product = col_diagonal_product * col_current_number
                    max_product = max(max_product, col_diagonal_product)
                col_diagonal_count += 1

    # Left diagonal
    for start_index in range(3, grid_size):
        row_diagonal_product = 1
        row_diagonal_count = 1
        col_diagonal_product = 1
        col_diagonal_count = 1
        # Loop through all items in the diagonal starting at this point in the row
        for other_index in range(start_index):
            row_current_number = grid[start_index - other_index][other_index]
            if row_current_number == 0:
                row_diagonal_product = row_diagonal_count = 1
            else:
                if row_diagonal_count < 4:
                    row_diagonal_product = row_diagonal_product * row_current_number
                else:
                    if ( (0 <= start_index - other_index + 4 < grid_size) and
                            (0 <= other_index - 4 < grid_size) and
                            (grid[start_index - other_index + 4][other_index - 4] != 0)):
                        row_diagonal_product = row_diagonal_product // grid[start_index - other_index + 4][other_index - 4]
                    row_diagonal_product = row_diagonal_product * row_current_number
                    max_product = max(max_product, row_diagonal_product)
                row_diagonal_count += 1

            col_current_number = grid[other_index][start_index - other_index]
            if col_current_number == 0:
                col_diagonal_product = col_diagonal_count = 1
            else:
                if col_diagonal_count < 4:
                    col_diagonal_product = col_diagonal_product * col_current_number
                else:
                    if ( (0 <= start_index - other_index + 4 < grid_size) and
                            (0 <= other_index - 4 < grid_size) and
                            (grid[other_index - 4][start_index - other_index + 4] != 0)):
                        col_diagonal_product = col_diagonal_product // grid[other_index - 4][start_index - other_index + 4]
                    col_diagonal_product = col_diagonal_product * col_current_number
                    max_product = max(max_product, col_diagonal_product)
                col_diagonal_count += 1

    return max_product
```
