```python
from Util.BigInt import make_bigint
from Util.Problems import Problem, solution


class P020(Problem):
    number = 20
    title = "Factorial Digit Sum"
    description = """<p>$n!$ means $n \\times (n - 1) \\times \\cdots \\times 3 \\times 2 \\times 1$.</p><p>For example, $10! = 10 \\times 9 \\times \\cdots \\times 3 \\times 2 \\times 1 = 3628800$,<br/>and the sum of the digits in the number $10!$ is $3 + 6 + 2 + 8 + 8 + 0 + 0 = 27$.</p><p>Find the sum of the digits in the number $100!$.</p>"""
```

```python
p = P020()
p.describe()
```

### Solution notes
I made a numba-compatible bigint class for doing calculations with very big integer numbers. While it is functional, it was a lot of work, which sadly does not yield any results yet (see next solution).

```python
@solution(P020, first=True, make_fast=True)
def bigint_brute_force():
    factorial_result = make_bigint("1")
    for n in range(2, 101):
        factorial_result *= n
    return sum(factorial_result.value)
```

```python
p.test_all()
```

Sadly, default python without numba can calculate $100!$ much faster than my bigint class.

```python
@solution(P020, make_fast=False)
def python_brute_force():
    factorial_result = 1
    for n in range(2, 101):
        factorial_result *= n
    sum_of_digits = 0
    while factorial_result:
        sum_of_digits += factorial_result % 10
        factorial_result //= 10
    return sum_of_digits
```

```python
p.test_all()
```

As we are counting the final digits, multiplying by 10 is unnecessary, as it only results in an extra 0 at the end of the final result. Therefore, if the number we are multiplying by is divisible by 10, we multiply by the value divided y 10 instead. This does not seem to speed up the algorithm very much.

```python
@solution(P020, best=True, make_fast=False)
def no_multiples_of_ten():
    factorial_result = 1
    for n in range(2, 101):
        if n % 10 == 0:
            factorial_result *= (n // 10)
        else:
            factorial_result *= n
    sum_of_digits = 0
    while factorial_result:
        sum_of_digits += factorial_result % 10
        factorial_result //= 10
    return sum_of_digits
```

```python
p.test_all()
```
