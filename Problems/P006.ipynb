```python
from Util.Problems import Problem, solution


class P006(Problem):
    number = 6
    title = "Sum Square Difference"
    description = """<p>The sum of the squares of the first ten natural numbers is,</p>
$$1^2 + 2^2 + ... + 10^2 = 385.$$
<p>The square of the sum of the first ten natural numbers is,</p>
$$(1 + 2 + ... + 10)^2 = 55^2 = 3025.$$
<p>Hence the difference between the sum of the squares of the first ten natural numbers and the square of the sum is $3025 - 385 = 2640$.</p>
<p>Find the difference between the sum of the squares of the first one hundred natural numbers and the square of the sum.</p>"""
    n = 100
```

```python
p = P006()
p.describe()
```

### Solution notes
Very simple, we just add all squares and all numbers and then square the latter and subtract the sum of squares.

```python
@solution(P006, first=True, make_fast=True, warmup_args=(P006.n, ))
def brute_force(n):
    sum_of_squares = 0
    sum_of_numbers = 0
    for i in range(1, n + 1):
        sum_of_numbers += i
        sum_of_squares += i ** 2
    sum_square_difference = (sum_of_numbers ** 2) - sum_of_squares
    return sum_square_difference
```

```python
p.test_all()
```

This problem is so basic, I'm not yet tempted to try for optimisation. Might look at it again later, but as it is taking less than a microsecond to solve, there is not much to be gained.

```python
@solution(P006, best=True, make_fast=True, warmup_args=(P006.n, ))
def best_solve(n):
    return brute_force(n)
```
