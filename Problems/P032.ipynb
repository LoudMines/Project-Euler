```python
from Util.Problems import Problem, solution

import math

class P032(Problem):
    number = 32
    title = "Pandigital Products"
    description = """<p>We shall say that an $n$-digit number is pandigital if it makes use of all the digits $1$ to $n$ exactly once; for example, the $5$-digit number, $15234$, is $1$ through $5$ pandigital.</p><p>The product $7254$ is unusual, as the identity, $39 \\times 186 = 7254$, containing multiplicand, multiplier, and product is $1$ through $9$ pandigital.</p><p>Find the sum of all products whose multiplicand/multiplier/product identity can be written as a $1$ through $9$ pandigital.</p><div class="note">HINT: Some products can be obtained in more than one way so be sure to only include it once in your sum.</div>"""
```

```python
p = P032()
p.describe()
```

### Solution notes
With a total of 9 digits, the sum can either be written as $d_1 \times d_2d_3d_4d_5 = d_6d_7d_8d_9$ or as $d_1d_2 \times d_3d_4d_5 = d_6d_7d_8d_9$

Therefore, we loop through all numbers 2-99 which do not have repeating digits (as we are staying below 100, this means they cannot be divisible by 11) nor do they contain the number 0. For each of these multiplicands, we go through each possible 4-digit number from the remaining digits and take it to be our product. We then divide the 4-digit number by our multiplicand. If the resulting multiplier consist of the same digits as the remaining digits, this is a solution to this problem. We save the product to a list of results, and finally return the sum of that list.


```python
@solution(P032, first=True, make_fast=True)
def brute_force():
    def digits_of(n):
        digits = set()
        while n > 0:
            digits.add(n % 10)
            n //= 10
        return digits
    existing_digits = set()
    for d in range(1, 10):
        existing_digits.add(d)
    results = []
    for multiplicand in range(2, 100):
        unique_digits = digits_of(multiplicand)
        if 0 in unique_digits or multiplicand % 11 == 0:
            continue

        usable_digits = existing_digits - unique_digits

        for thousand in usable_digits:
            thousands = 1000 * thousand
            no_thousands_digits = usable_digits - {thousand}
            for hundred in no_thousands_digits:
                hundreds = thousands + hundred * 100
                no_hundreds_digits = no_thousands_digits - {hundred}
                for ten in no_hundreds_digits:
                    tens = hundreds + ten * 10
                    no_tens_digits = no_hundreds_digits - {ten}
                    for one in no_hundreds_digits:
                        product = tens + one
                        if product % multiplicand != 0 or product in results:
                            continue
                        remaining_digits = no_tens_digits - {one}
                        multiplier = product // multiplicand
                        multiplier_digits = digits_of(multiplier)
                        if multiplier_digits == remaining_digits:
                            results.append(product)

    return sum(results)
```

```python
p.test_once("brute_force")
```

```python
@solution(P032, best=True, make_fast=True)
def best_solve():
    return brute_force()
```
